# Kalman Filters from Scratch

**A minimal, self-contained walk-through** of the discrete-time Kalman filter — derive the
predict and update steps, and see how the gain balances **measurement noise** $\mathcal{R}$
against **process noise** $\mathcal{Q}$.

## 1 · The linear Gaussian state-space model

$x_t = F x_{t-1} + w_t, \qquad w_t \sim \mathcal{N}(0, Q)$$
$z_t = H x_t + v_t, \qquad v_t \sim \mathcal{N}(0, R)$$

```python
import numpy as np

F = np.array([[1.0, 1.0], [0.0, 1.0]])
H = np.array([[1.0, 0.0]])
Q = np.eye(2) * 0.01
R = np.array([[1.0]])

def kf_step(x, P, z):
    # predict
    x = F @ x
    P = F @ P @ F.T + Q
    # update
    S = H @ P @ H.T + R
    K = P @ H.T @ np.linalg.inv(S)
    x = x + K @ (z - H @ x)
    P = (np.eye(2) - K @ H) @ P
    return x, P

x = np.array([0.0, 1.0]); P = np.eye(2)
x, P = kf_step(x, P, np.array([1.2]))
print("Kalman gain K =", K[0])
```

## 2 · Why the gain matters

- If $R \gg Q$ (noisy measurements), the filter **trusts the model**; $K \to 0$.
- If $Q \gg R$ (unreliable dynamics), the filter **trusts the measurement**; $K \to 1$.
